# Ejercicio 1

En una regresión lineal simple con intercepto:

    y_i = alpha + beta x_i + u_i

el estimador MCO de la pendiente beta mide cuánto cambia y, en promedio, ante
un cambio unitario en x.

La fórmula del estimador de la pendiente es:

    beta_hat = sum((x_i - x_bar)(y_i - y_bar)) / sum((x_i - x_bar)^2)

El numerador mide cómo se mueven conjuntamente x e y respecto de sus medias,
es decir, la covarianza muestral multiplicada por n-1:

    Cov(x,y) = sum((x_i - x_bar)(y_i - y_bar)) / (n-1)

El denominador mide la dispersión de x respecto de su media, es decir, la
varianza muestral de x multiplicada por n-1:

    Var(x) = sum((x_i - x_bar)^2) / (n-1)

Entonces:

    beta_hat =
    [sum((x_i - x_bar)(y_i - y_bar)) / (n-1)]
    ------------------------------------------------
    [sum((x_i - x_bar)^2) / (n-1)]

Como el factor (n-1) aparece tanto arriba como abajo, se cancela. Por lo tanto:

    beta_hat = Cov(x,y) / Var(x)

Aplicado a finanzas, si y es la prima de riesgo de un activo y x es la prima de
riesgo del mercado, el beta del activo es la covarianza muestral entre la prima
de riesgo del activo y la prima de riesgo del mercado, dividida por la varianza
muestral de la prima de riesgo del mercado.

Por eso, el beta no es solamente una pendiente de regresión: también puede
interpretarse como una covarianza normalizada por el riesgo del mercado.

Para testear si beta es igual a 1, planteo:

H0: beta = 1

HA: beta != 1

El p-value de Excel no sirve directamente en este caso, porque Excel reporta el
test contra beta = 0. Por eso calculo manualmente el estadístico:

t = (beta_hat - 1) / SE(beta_hat)

t = (0.69758673 - 1) / 0.14933314 = -2.0251

gl = n - cantidad de parámetros estimados

gl = 59 - 2

gl = 57


Con 57 grados de libertad, el p-value a dos colas es aproximadamente 0.0475.

Como el p-value es menor que 0.05, rechazo H0 al 5%. Por lo tanto, hay evidencia
estadística para decir que el beta de IBM es distinto de 1. Además, como el beta
estimado es menor que 1, IBM parece tener una sensibilidad al mercado menor que
la cartera de mercado.

```python
beta_hat = 0.69758673
se_beta = 0.14933314
beta_0 = 1

t_stat = (beta_hat - beta_0) / se_beta
t_stat
```

```python
gl = 57

p_value = 2 * (1 - stats.t.cdf(abs(t_stat), df=gl))

t_stat, p_value
```

```python
ruta = "/Users/bautistagianfrancisco/Library/Mobile Documents/com~apple~CloudDocs/MIA 2026/2 Trim/Modelado Estocastico/Bases de Datos MIA103/MIA103 Ejercitación 3 Datos.xlsx"

ceo = pd.read_excel(
    ruta,
    sheet_name="Datos_Ejer_3_CEO"
)

ceo.head()
```

```python
ceo.columns
```

```python
ceo.columns = ["profits", "salary"]

ceo.head()
```

```python

y = ceo["salary"]      # compensacion CEO
x = ceo["profits"]     # ganancia empresa

x = sm.add_constant(x)

modelo_ceo = sm.OLS(y, x).fit()

print(modelo_ceo.summary())
```

La pendiente estimada es aproximadamente 0.0008. Esto significa que, por cada
aumento de 1 millón de dólares en las ganancias de la empresa, la compensación
esperada del CEO aumenta en promedio 0.0008 millones de dólares.

Como la compensación también está medida en millones de dólares, 0.0008 millones
equivalen aproximadamente a 800 dólares.

Por lo tanto, un aumento de 1 millón de dólares en las ganancias de la empresa
se asocia con un aumento promedio de alrededor de 800 dólares en la compensación
del CEO.

```python
beta_hat = modelo_ceo.params["profits"]
t_stat = modelo_ceo.tvalues["profits"]
p_value = modelo_ceo.pvalues["profits"]

print("Beta estimado:", beta_hat)
print("t-stat:", t_stat)
print("p-value:", p_value)
```

```python
alpha = 0.01

p_value = modelo_ceo.pvalues["profits"]

if p_value < alpha:
    print("Rechazo H0: la pendiente es estadísticamente distinta de cero al 1%.")
else:
    print("No rechazo H0: no hay evidencia suficiente al 1%.")
```

Para testear si las ganancias de la empresa tienen una relación estadísticamente
significativa con la compensación del CEO, planteo:

H0: beta = 0

HA: beta != 0

donde beta es la pendiente asociada a profits.

Según la regresión, el estadístico t de la pendiente es aproximadamente 7.228 y
el p-value es aproximadamente 0.000, menor al nivel de significancia del 1%.

Por lo tanto, rechazo la hipótesis nula. Hay evidencia estadística suficiente
para afirmar que, en esta muestra, mayores ganancias de la empresa se asocian
con mayor compensación al CEO.

```python
r2 = modelo_ceo.rsquared

print("R²:", r2)
print("Porcentaje explicado:", r2 * 100)
```

El porcentaje de la variabilidad en las compensaciones a los CEOs explicado por
las ganancias de las empresas está dado por el R² de la regresión.

En este caso, el R² es aproximadamente 0.434. Esto significa que alrededor del
43.4% de la variabilidad observada en la compensación de los CEOs puede ser
explicada por las ganancias de las empresas.

El porcentaje restante, aproximadamente 56.6%, queda explicado por otros factores
que no están incluidos en esta regresión simple y por el término de error.
